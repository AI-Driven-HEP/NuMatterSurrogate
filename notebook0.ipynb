```python
import sys
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')


ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT))


plt.rcParams['text.usetex']=True
plt.style.use('classic')
```

```python
data = np.load(ROOT / 'Data' / "parameter_samples.npz")


param = data["param"]

eta = data["eta"]
energy = data["energy"]

U_evolved = data["U_Evol"]

mass_weights = data["Mass_W"]

print("Loaded:")
print("param :", param.shape)
print("eta     :", eta.shape)
print("energy  :", energy.shape)
print("U_evolved  :", U_evolved.shape)
print("mass_weights   :", mass_weights.shape)
```

```python
index = 30

print("theta12 :", param[index,0])
print("dm21    :", param[index,1])
```

```python
plt.plot(energy, np.sum(U_evolved[index, -1] * mass_weights[index], axis=1), lw =2)
plt.plot(energy, np.sum(U_evolved[index, 0] * mass_weights[index], axis=1), lw =2)
```

```python
image = np.sum(U_evolved[index] * mass_weights[index], axis=2)



plt.imshow(
    image,
    origin='lower',
    extent=[1, 20, 0, 110],
    aspect='auto',
)

plt.xlabel(r'$E_\nu$ [MeV]')
plt.ylabel(r'$\eta$ [degrees]')
plt.colorbar(label='P_ee')
plt.show()
```
